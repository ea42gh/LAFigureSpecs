# Formatting Helpers and Renderer Entrypoints

This notebook covers top-level formatting helpers and high-level render entrypoints.


```python
import sympy as sym
import LAFigureSpecs as lf
```

## Formatting helpers

- `make_decorator`, `decorate_tex_entries`
- selectors: `sel_entry`, `sel_row`, `sel_col`, `sel_box`, `sel_diag`, `sel_antidiag`, `sel_mask`


```python
dec = lf.make_decorator(text_color="red", bf=True)
print(dec("x"))
```

## High-level renderer entrypoints

- `ge_tex/svg`
- `qr_tex/svg`
- `eig_tex/svg`

Use these when you want LAFigureSpecs to build the figure spec and render it. Lower-level matrixlayout renderers are renderer internals.


```python
A = sym.Matrix([[1, 2], [3, 4]])
tex = lf.ge_tex(A)
svg = lf.ge_svg(A)
print(type(tex), type(svg))
```
